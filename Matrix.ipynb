```python
import numpy as np
import numpy.linalg as LA
```

```python
A = np.array([[2, -1], [0, 3]])
B= np.array([[1, 2], [-3, 4]])
# A + B
add_result = A + B
# AB
mul_result = A @ B
# 3A - 2B
exp_result = 3*A - 2*B
print("A + B:\n", add_result)
print("AB:\n", mul_result)
print("3A - 2B:\n", exp_result)
```

In this code we utilized the @ to do matrix multiplication and + to do addition and * for scalar multiplication

```python
detA = LA.det(A)
print("Determinant of A:", detA)

if detA != 0:
    invA = LA.inv(A)
    print("Matrix A is invertible and")
    print("Inverse of A:\n", invA)
else:
    print("Matrix A is not invertible")
```

First we utilized a function of numpy called det to find determinant of the matrix and we checked if the determinant is non zero, we utilized the function of inv form numpy to get inverse of matrix

```python
C = np.array([[4, 1],
              [-2, 1]])
print("Matrix A:\n", A)
eigenvalues, eigenvectors = LA.eig(A)
print("Eigenvalue:")
print(eigenvalues[0], eigenvalues[1])
print("Corresponding Eigenvector:")
print(eigenvectors)
```

Here we utilized the numpy function of eig in order to get eigen values and vectors

```python
x = np.array([[4, 1.5, 100],
                [6, 2.0, 150]])
y = np.array([20, 30])
xtranspose = x.T
beta = LA.pinv(xtranspose @ x) @ xtranspose @ y
print("Coefficient vector(beta):\n", beta)
xnew = np.array([8, 3.0, 200])
predictedprice = xnew @ beta
print("Predicted price of car 3:", predictedprice)
```

Here we utilized the aspect of vector and its position to predict the price of the third car where we utilized pinv to find the pseudoinverse as normal inverse is not applicable for all tyoe of matrix.
