```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

Here pandas is imported as pd.It is a python library which is used for data manipulation and analysis of datasets.
Seaborn is imported as sns.It is used for statistical and data analysis and it is primarily utilized to make the look better by default.
Here matplotlib.pyplot is imported as mat used for creating graph and charts.


```python
def load_dataset(file_path):
  return pd.read_csv(file_path)
df = load_dataset("/content/Health_heart_experimental.csv")
df.head(40)
```

Here we loaded a data set under the name of df.
It took data from health_heart_experimental.csv and utilized it to get 40 set of data regarding the condition of the patients

```python
df_line = df.groupby('age')['SysBP'].mean().reset_index()

plt.figure()
plt.plot(df_line['age'], df_line['SysBP'], color='green')

plt.title("Average Systolic Blood Pressure vs Age")
plt.xlabel("Age")
plt.ylabel("Average Systolic Blood Pressure")
plt.show()
```

The above code utilized the functions from matplotlib.pyplot such as xlabel , ylabel , plot ,figure ,show etc.It  utilized it to show the data of systolic blood pressure and age in a line chart.
From this line chart,it  can be seen that as human get older they start to have higher blood pressure around the age of 60 due to their body degrading itself.it starts to cause more strain in thier arteries and can severe damage and it often refrerred as a silent killer.

```python
df_line = df.groupby('DiaBP')['SysBP'].mean().reset_index()

plt.figure()
plt.plot(df_line['DiaBP'], df_line['SysBP'], color='green')

plt.title("Average Systolic Blood Pressure vs Average diastolic Blood Pressure")
plt.xlabel("Average diastolic Blood Pressure")
plt.ylabel("Average Systolic Blood Pressure")
plt.show()
```

By utilizing the matplotlib function we made a comparision between average systolic blood pressure and diastolic blood pressure.From the chart we can find about the pulse pressure between the patients from which we can determine the condition of their arteries. we can see that in the chart that the people below the dystolic range of 100 ten d to have a normal amount of blood pluse whereas above that the people seems to have low blood pulse which indicates a heart issues or low stroke volume.

```python
counts = df['indication'].value_counts()

plt.bar(counts.index.astype(str), counts.values)
plt.title("Heart Disease Indication Count")
plt.xlabel("Indication (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.show()
```

Here a chart of heart disease indication count is used to compare the count the number of indication of people with heart disease and it is shown that the count for no indication is slightly higher than indication  for heart disease which can tell us that nearlry 45% of people have some heart disease indications but the people with indication may not all have heart disease as the global rate for heart disease is still 16% which is significantly lower than 45% indicating that major part of those with heart disease will still not heart disease.

```python
df['sex'].value_counts().plot(kind='pie', autopct='%0.2f')
plt.title("Sex Distribution")
plt.ylabel("")
plt.show()
```

here we created a pie chart in order to reperesent the sex distribution where 1 is female and 0 is male. From this we can conclude that nearly both gender have participoated equally for the formation of this dataset and can be used to find the specific issues of heart disese for each genders

```python
sns.boxplot(x='sex',y='age',hue='indication',data=df)
plt.title("Box Plot of age,sex and indications")
plt.show()
```

From the above table we utilized box plot to find out the oultlier inn age,sex and indication but since it can seen that regardless of age or sex all are susceptible to heart disease and we can see that gender are not a factor for heart disease as the median for both gender are nearly equal.

```python
sns.histplot(data=df, x='age', hue='indication',multiple='stack')
plt.title('Age Distribution by Indication')
plt.show()
```

we utilized a histogram to plot a graph to find te age distribution between the peope who participated in this experiment for data.we found that some age groups took significantly more than other but regardless of that it seems that heart dieases indication major factor is not age as nealy in all age as the ratio of no indication and indication is relatively same.

```python
sns.displot(data=df, x='weightKg', hue='indication', kind='kde')
plt.title('weight Distribution by Indication')
plt.show()
```

Here we drew a histogram of type Kde of weight distribution by Indication and found out that the people with weight over  100 kg neaky all have indication of for heart diseases which coincides with scientific studies that overweight people have higher chances of heart diseases. It also indicates that people under the weight of 60 have also have higher risk of heart disease due to being under weight while the weight if 60 is the most optimal with significantly better ratio of no indication and indication.

```python
df_scatter = df.groupby(['age', 'sex'])['SysBP'].mean().reset_index()

plt.figure()
sns.scatterplot(
    data=df_scatter,
    x='age',
    y='SysBP',
    hue='sex',
    s=100
)

plt.title("Average Systolic Blood Pressure by Age and Sex")
plt.xlabel("Age")
plt.ylabel("Average Systolic Blood Pressure")
plt.show()
```

From the above scatterplot it can be concluded that the that as age is increased the mean blood pressure rises slightly signifying the major reasons for high systolic blood pressure is usually their diet,lifestlye etc.
